```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

media = 70
desv_std = 4
total_estudiantes = 500
x_curva = np.linspace(media - 4*desv_std, media + 4*desv_std, 1000)
y_curva = norm.pdf(x_curva, media, desv_std)

limite_68 = 68
limite_72 = 72

prob_entre = norm.cdf(limite_72, media, desv_std) - norm.cdf(limite_68, media, desv_std)
cant_entre = total_estudiantes * prob_entre

print(f"Distribución Normal: Estudiantes ITD")
print(f"Probabilidad 68 < peso < 72 kg: {prob_entre:.4f}")
print(f"Estudiantes entre 68 y 72 kg: {cant_entre:.2f}\n")

fig, axs = plt.subplots(1, 1, figsize=(8, 5))

axs.plot(x_curva, y_curva, 'b')
axs.fill_between(x_curva, y_curva, where=(x_curva >= limite_68) & (x_curva <= limite_72), color='skyblue', alpha=0.5)
axs.set_title('Probabilidad 68 < Peso < 72 kg')
axs.grid(True)

plt.tight_layout()
plt.show()
```
