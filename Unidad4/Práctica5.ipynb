```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

plt.figure(figsize=(6, 4))

media = 70
desv_std = 4
total_estudiantes = 500

peso_inf = 68
peso_sup = 72

prob_68 = norm.cdf(peso_inf, media, desv_std)
prob_72 = norm.cdf(peso_sup, media, desv_std)
prob_entre_68_72 = prob_72 - prob_68

print("probabilidad de que un estudiante pese entre 68 y 72 kg:", prob_entre_68_72)


x = np.linspace(media - 4 * desv_std, media + 4 * desv_std, 1000)
y = norm.pdf(x, media, desv_std)

plt.plot(x, y, 'b', linewidth=2)
plt.fill_between(x, y, where=((x >= peso_inf) & (x <= peso_sup)), color='skyblue', alpha=0.5)
plt.title('probabilidad de que un estudiante pese entre 68 y 72 kg.\n')
plt.xlabel('peso (kg)')
plt.ylabel('D+densidad de probabilidad')
plt.grid(True)
plt.show()
```

```python
plt.figure(figsize=(6, 4))

media = 70
desv_std = 4
total_estudiantes = 500


peso_inf = 68
peso_sup = 72


prob_68 = norm.cdf(peso_inf, media, desv_std)
prob_72 = norm.cdf(peso_sup, media, desv_std)
prob_entre_68_72 = prob_72 - prob_68


estudiantes_entre_68_72 = total_estudiantes * prob_entre_68_72

print("número de estudiantes que pesan entre 68 y 72 kg:", estudiantes_entre_68_72)


x = np.linspace(media - 4 * desv_std, media + 4 * desv_std, 1000)
y = norm.pdf(x, media, desv_std)

plt.plot(x, y, 'b', linewidth=2)
plt.fill_between(x, y, where=((x >= peso_inf) & (x <= peso_sup)), color='skyblue', alpha=0.5)
plt.title('número de estudiantes con peso entre 68 y 72 kg.\n')
plt.xlabel('peso (kg)')
plt.ylabel('densidad de probabilidad')
plt.grid(True)
plt.show()
```
