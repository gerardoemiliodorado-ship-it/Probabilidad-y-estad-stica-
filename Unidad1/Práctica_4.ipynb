```python
import random
import matplotlib.pyplot as plt
from statistics import *

fabrica = []
for i in range(100):
  numero_aleatorio = round(random.uniform(50,100), 2)
  fabrica.append(numero_aleatorio)
  print(fabrica)
```

```python
fabrica2 = []
for i in range(100):
  numero_aleatorio = round(random.uniform(70,100), 2)
  fabrica2.append(numero_aleatorio)
  print(fabrica2)
```

```python
plt.boxplot(fabrica)
plt.ylabel("producción en mts.")
plt.title("producción de la fabrica")
plt.show()
```

```python
fig, (ax1, ax2) = plt.subplots(nrows=1, ncols=2, figsize=(9, 4))
bplot1 = ax1.boxplot(fabrica, vert=True, patch_artist=True)
bplot2 = ax2.boxplot(fabrica2, vert=True, patch_artist=True)
ax1.set_title('producción de la fabrica')
ax2.set_title('producción de la fabrica 2')
plt.show()
```

```python
fabrica.append(150)
print(fabrica)
```

```python
fig, (ax1, ax2) = plt.subplots(nrows=1, ncols=2, figsize=(9, 5))
bplot1 = ax1.boxplot(fabrica, vert=True, patch_artist=True)
ax1.set_ylim(30,160)
bplot2 = ax2.boxplot(fabrica2, vert=True, patch_artist=True)
ax2.set_ylim(30,160)
ax1.set_title('producción de la fabrica')
ax2.set_title('producción de la fabrica 2')
plt.show()
```
