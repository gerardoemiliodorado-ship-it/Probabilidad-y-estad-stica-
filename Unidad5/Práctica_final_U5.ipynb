```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.linear_model import LinearRegression
from scipy.stats import pearsonr
from IPython.display import display

sns.set(style="whitegrid")


def ecuacion(a, b, dec=2):
    signo = "+" if b >= 0 else "-"
    return f"Ŷ = {a:.{dec}f} {signo} {abs(b):.{dec}f}X"


def analizar_grupo(n):
    print(f"Grupo {n}")

    print(f"\n{n}.1 Carga y exploración inicial\n")
    df = pd.read_csv(f"grupo{n}.csv")
    display(df.head())

    print("\nFilas y columnas:", df.shape, "\n")
    df.info()
    print("\nValores nulos por columna:")
    print(df.isnull().sum())

    print("\nEstadísticos descriptivos:")
    display(df.describe())

    plt.figure(figsize=(7, 4))
    sns.histplot(df["Calificacion"], bins=10, kde=True, color="steelblue")
    plt.title(f"Histograma de Calificaciones Finales - Grupo {n}")
    plt.xlabel("Calificación Final")
    plt.ylabel("Frecuencia")
    plt.show()

    plt.figure(figsize=(7, 4))
    sns.scatterplot(x="HorasEstudio", y="Calificacion", data=df, color="blue")
    plt.title(f"Diagrama de Dispersión: Horas de Estudio vs Calificación - Grupo {n}")
    plt.xlabel("Horas de Estudio")
    plt.ylabel("Calificación Final")
    plt.grid(True)
    plt.show()

    print(f"\n{n}.2 Modelo de regresión lineal simple\n")
    X = df[["HorasEstudio"]]
    y = df["Calificacion"]

    modelo = LinearRegression()
    modelo.fit(X, y)

    b = modelo.coef_[0]
    a = modelo.intercept_
    r2_modelo = modelo.score(X, y)

    print(f"Pendiente (b): {b:.4f}")
    print(f"Intercepto (a): {a:.4f}")
    print(f"R² del modelo: {r2_modelo:.4f}")
    print(f"Ecuación de regresión: {ecuacion(a, b, 4)}")

    horas = 10
    pred = modelo.predict(pd.DataFrame({"HorasEstudio": [horas]}))
    print(f"Calificación estimada con {horas} horas de estudio: {pred[0]:.2f}")

    print(f"\n{n}.3 Correlación de Pearson\n")
    r, p = pearsonr(df["HorasEstudio"], df["Calificacion"])
    print(f"Coeficiente de correlación de Pearson (r): {r:.4f}")
    print(f"Valor p: {p:.4e}")
    print(f"Coeficiente de determinación (r²): {r**2:.4f}")

    print(f"\n{n}.4 Visualización: datos y recta de regresión\n")
    y_pred = modelo.predict(X)

    plt.figure(figsize=(8, 5))
    plt.scatter(df["HorasEstudio"], df["Calificacion"], color="blue", label="Datos")
    plt.plot(df["HorasEstudio"], y_pred, color="red",
             label=f"Línea de regresión: {ecuacion(a, b)}")
    plt.title(f"Regresión Lineal: Horas de Estudio vs Calificación Final - Grupo {n}")
    plt.xlabel("Horas de Estudio")
    plt.ylabel("Calificación Final")
    plt.legend(title=f"r = {r:.3f}")
    plt.grid(True)
    plt.show()

    plt.figure(figsize=(8, 5))
    sns.regplot(x="HorasEstudio", y="Calificacion", data=df,
                scatter_kws={"color": "blue"}, line_kws={"color": "red"})
    plt.title(f"Diagrama de Dispersión con Línea de Regresión (seaborn) - Grupo {n}")
    plt.xlabel("Horas de Estudio")
    plt.ylabel("Calificación Final")
    plt.show()

    return df, modelo, X, a, b, r, p


df1, modelo1, X1, a1, b1, r1, p1 = analizar_grupo(1)
df2, modelo2, X2, a2, b2, r2, p2 = analizar_grupo(2)
df3, modelo3, X3, a3, b3, r3, p3 = analizar_grupo(3)

print("Resumen comparativo\n")

resumen = pd.DataFrame({
    "Grupo": ["Grupo 1", "Grupo 2", "Grupo 3"],
    "Intercepto (a)": [a1, a2, a3],
    "Pendiente (b)": [b1, b2, b3],
    "r (Pearson)": [r1, r2, r3],
    "r²": [r1**2, r2**2, r3**2],
}).round(4)
resumen["Valor p"] = [f"{p:.2e}" for p in (p1, p2, p3)]
resumen["Ecuación"] = [ecuacion(a1, b1), ecuacion(a2, b2), ecuacion(a3, b3)]
display(resumen)

fig, ejes = plt.subplots(1, 3, figsize=(15, 4.5), sharey=True)
datos = [(df1, modelo1, X1, "Grupo 1", r1),
         (df2, modelo2, X2, "Grupo 2", r2),
         (df3, modelo3, X3, "Grupo 3", r3)]

for eje, (df, modelo, X, nombre, r) in zip(ejes, datos):
    eje.scatter(df["HorasEstudio"], df["Calificacion"], color="blue", label="Datos")
    eje.plot(df["HorasEstudio"], modelo.predict(X), color="red", label="Línea de regresión")
    eje.set_title(f"{nombre} (r = {r:.3f})")
    eje.set_xlabel("Horas de Estudio")
    eje.grid(True)
    eje.legend()

ejes[0].set_ylabel("Calificación Final")
plt.suptitle("Comparación de la regresión lineal en los tres grupos")
plt.tight_layout()
plt.show()

# Aún faltan las conclusiones y las preguntas
```

## Interpretación de resultados

### Grupo 1

**1. Coeficiente de regresión:** la ecuación es Ŷ = 85.31 - 0.11X, la pendiente es casi cero, o sea que estudiar más horas no cambia la calificación en este grupo

**2. Correlación:** r = -0.057, es negativa pero prácticamente nula, no hay relación lineal y en la gráfica los datos se ven regados, el r² es 0.003 así que las horas explican solo el 0.3% de las calificaciones

**3. ¿Es significativo?** no, el valor p es 0.69 y es mayor que 0.05, lo poco que se ve podría ser casualidad

**4. Confiabilidad:** no es confiable, el modelo daría casi la misma calificación (unos 84) sin importar las horas, creo que aquí influyen otros factores y no solo el tiempo de estudio

### Grupo 2

**1. Coeficiente de regresión:** la ecuación es Ŷ = 15.31 + 4.22X, por cada hora extra de estudio la calificación sube en promedio 4.22

**2. Correlación:** r = 0.862, es positiva y fuerte, a más horas mayor calificación, el r² es 0.743 así que las horas explican cerca del 74% de las calificaciones, hay datos que se salen de la recta pero son atípicos normales

**3. ¿Es significativo?** sí, el valor p es 9.1e-16 y es mucho menor que 0.05

**4. Confiabilidad:** sirve para dar una idea general, pero no para una calificación exacta, hay casos reales que se alejan bastante de la recta y solo funciona dentro del rango de horas de los datos (4 a 20)

### Grupo 3

**1. Coeficiente de regresión:* la ecuación es Ŷ = 0.11 + 5.29X, por cada hora extra la calificación sube en promedio 5.29, pero no me parece realista como regla general porque cerca de las 17 a 20 horas las calificaciones se aplanan en 100

**2. Correlación:** r = 0.942, es positiva y muy fuerte, la más alta de los tres grupos, el r² es 0.888 así que las horas explican cerca del 89% de las calificaciones

**3. ¿Es significativo?** sí, el valor p es 1.9e-24, prácticamente cero y mucho menor que 0.05

**4. Confiabilidad:** no le tendría mucha confianza, aunque la correlación es muy fuerte el modelo falla con muchas horas, por ejemplo con 25 horas daría más de 100, sirve más como aproximación que como predicción exacta

## Conclusiones generales

Los tres grupos dieron resultados muy distintos con las mismas variables, en el grupo 1 no hay relación (r = -0.057), en el grupo 2 es fuerte (r = 0.862) y en el grupo 3 es muy fuerte (r = 0.942), con la misma cantidad de alumnos hay grupos donde se ve un patrón claro y otros donde no, por eso los datos importan y hay que analizarlos bien antes de concluir, también un r alto no significa que el modelo sea perfecto, como en el grupo 3 que falla con muchas horas porque la calificación tiene tope de 100, en general estudiar más sí suele ayudar pero no siempre lo garantiza, como se ve en el grupo 1
